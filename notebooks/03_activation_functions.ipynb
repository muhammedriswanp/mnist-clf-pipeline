```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def tanh(z):
    return np.tanh(z)

x = np.linspace(-6, 6, 300)
```

```python
fig, ax = plt.subplots(figsize=(10, 4))

ax.plot(x, relu(x),      label='ReLU',       color='steelblue',  linewidth=2)
ax.plot(x, sigmoid(x),   label='Sigmoid',    color='tomato',     linewidth=2)
ax.plot(x, tanh(x),      label='Tanh',       color='seagreen',   linewidth=2)

ax.set_title("All Activation Functions — Comparison")
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.show()
```
