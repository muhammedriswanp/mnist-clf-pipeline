```python
import torch
import matplotlib.pyplot as plt
import numpy as np
from torchvision import datasets, transforms
```

```python
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))
])

dataset = datasets.MNIST(root='../data', train=True, download=False, transform=transform)

print(f"Total samples : {len(dataset)}")
image, label = dataset[0]

print(image.shape)
print(label)

print(f"Classes       : {dataset.classes}")
```

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
fig.suptitle("MNIST Sample Images", fontsize=14)

for i, ax in enumerate(axes.flat):
    img, label = dataset[i]
    ax.imshow(img.squeeze(), cmap='gray') #img.squeeze() => removes dimensions with size 1. imshow expect [28, 28]
    ax.set_title(f"Label: {label}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
labels = [dataset[i][1] for i in range(len(dataset))]

fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(range(10), [labels.count(i) for i in range(10)], color='steelblue')
ax.set_xticks(range(10))
ax.set_xlabel("Digit Class")
ax.set_ylabel("Count")
ax.set_title("Class Distribution in MNIST Training Set")
plt.tight_layout()
plt.show()
```
