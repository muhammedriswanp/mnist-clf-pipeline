```python
import json
import matplotlib.pyplot as plt

with open("../evaluation/metrics_adam.json") as f:
    m = json.load(f)

epochs = range(1, len(m["train_losses"]) + 1)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(epochs, m["train_losses"], label="Train Loss")
ax1.plot(epochs, m["val_losses"],   label="Val Loss")
ax1.set_title("Loss Curve (Adam)")
ax1.set_xlabel("Epoch")
ax1.set_ylabel("Loss")
ax1.grid()
ax1.legend()

ax2.plot(epochs, m["train_accs"], label="Train Acc")
ax2.plot(epochs, m["val_accs"],   label="Val Acc")
ax2.set_title("Accuracy Curve (Adam)")
ax2.set_xlabel("Epoch")
ax2.set_ylabel("Accuracy")
ax2.grid()
ax2.legend()

plt.tight_layout()
plt.show()
```
